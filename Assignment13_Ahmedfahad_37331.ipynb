```python
import pandas as pd
from pathlib import Path

# Change this path if you run the notebook outside the provided workspace.
file_path = Path("teaching_prices_clean(1).csv")
df = pd.read_csv(file_path)
df["date"] = pd.to_datetime(df["date"])

print("Rows:", len(df))
print("Date range:", df["date"].min().date(), "to", df["date"].max().date())
print("Items:", sorted(df["item"].unique()))
print("Cities:", sorted(df["city"].unique()))
print("Columns:", df.columns.tolist())
```

```python
rows = []
for item, group in df.groupby("item"):
    first_date = group["date"].min()
    last_date = group["date"].max()
    first_avg = group.loc[group["date"].eq(first_date), "price"].mean()
    last_avg = group.loc[group["date"].eq(last_date), "price"].mean()
    rows.append({
        "item": item,
        "n": len(group),
        "mean": group["price"].mean(),
        "median": group["price"].median(),
        "first_date_average": first_avg,
        "last_date_average": last_avg,
        "first_to_last_change_pct": (last_avg / first_avg - 1) * 100
    })

item_stats = pd.DataFrame(rows).round(2)
item_stats
```

```python
city_rows = []
for (item, city), group in df.groupby(["item", "city"]):
    group = group.sort_values("date")
    first, last = group.iloc[0]["price"], group.iloc[-1]["price"]
    city_rows.append({
        "item": item, "city": city, "n": len(group),
        "mean": group["price"].mean(),
        "median": group["price"].median(),
        "first": first, "last": last,
        "change_pct": (last / first - 1) * 100
    })

city_stats = pd.DataFrame(city_rows).round(2)
city_stats
```
